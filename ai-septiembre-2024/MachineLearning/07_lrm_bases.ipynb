## Ejemplo de regresión lineal con Scikit Learn
### Parte 1: Obtención y limpieza de datos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
casas = pd.read_csv('USA_Housing.csv')
casas.head()
```

```python
casas.info() # Información del dataset
```

```python
casas.describe() # Descripción estadística
```

```python
casas.columns
```

```python
casas['Price']
```

```python
sns.histplot(casas['Price'], kde=True)
```

```python
# Seleccionar solo las columnas numéricas
casas_numericas = casas.select_dtypes(['float64', 'int64'])

# Calcular la matriz de correlación y generar el heatmap
sns.heatmap(casas_numericas.corr(), annot=True)
```

---
### Parte 2: División y entrenamiento del modelo

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
casas.head()
```

```python
casas.columns
```

```python
# Seleccionamos las características (variables explicativas) para el modelo
X = casas[['Avg. Area Income', 'Avg. Area House Age', 'Avg. Area Number of Rooms', 'Avg. Area Number of Bedrooms', 'Area Population']]
X.head()
```

```python
# Seleccionamos la variable dependiente (precio) para el modelo
y = casas['Price']
y.head()
```

```python
# Dividimos los datos en conjuntos de entrenamiento y prueba (70% entrenamiento, 30% prueba)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
# Creamos un modelo de regresión lineal y lo entrenamos
lrm = LinearRegression()
lrm.fit(X_train, y_train)
```

---
### Parte 3: Evaluación del modelo

```python
from sklearn import metrics
```

```python
# Realizamos predicciones utilizando el modelo entrenado (lrm) con los datos de prueba (X_test)
predicciones = lrm.predict(X_test)
predicciones
```

```python
y_test
```

```python
# Graficamos las predicciones frente a los valores reales de prueba
plt.scatter(y_test, predicciones)
```

```python
# Graficamos la distribución de los errores (diferencia entre los valores reales y las predicciones)
sns.histplot(y_test-predicciones, kde=True)
```

**MAE (Mean Absolute Error)**

```python
# Calculamos el error absoluto medio (MAE). 
# Este valor nos dice, en promedio, cuánto se equivocan nuestras predicciones respecto a los valores reales.
# El MAE se calcula sumando todas las diferencias absolutas entre los valores reales y las predicciones, 
# y luego dividiendo entre el número total de predicciones.

metrics.mean_absolute_error(y_test, predicciones)
```

**MSE (Mean Squared Error)**

```python
# Calculamos el error cuadrático medio (MSE). 
# Este valor es similar al MAE, pero en lugar de tomar la diferencia directa, toma el cuadrado de la diferencia.
# Esto hace que los errores más grandes tengan más peso en el cálculo, lo que es útil cuando se quiere penalizar errores grandes.
# El MSE se obtiene sumando los cuadrados de las diferencias entre los valores reales y las predicciones, 
# y luego dividiendo entre el número total de predicciones.

metrics.mean_squared_error(y_test, predicciones)
```

**RMSE (Root Mean Squared Error)**

```python
# Calculamos la raíz cuadrada del error cuadrático medio (RMSE).
# El RMSE es similar al MSE, pero tiene la ventaja de que está en las mismas unidades que los datos originales (mientras que MSE está en unidades al cuadrado).
# Nos da una idea de qué tan lejos están, en promedio, nuestras predicciones de los valores reales, 
# pero penaliza los errores grandes de manera más fuerte que el MAE.

np.sqrt(metrics.mean_squared_error(y_test, predicciones))
```
